```python
import pandas as pd


#base_path ="/scratch/usr/nimtsspi/"
base_path = "/mnt/vast-kisski/projects/kisski_tegami"
INPUT_PATH = f"{base_path}/datasets/TG"
OUTPUT_PATH = f"{base_path}/datasets/TG/labels"

```

```python
df = pd.read_parquet(f"{INPUT_PATH}/TG-weak-label-core.parquet")
```

## Gender Bias/Sexism

```python
gender_df = df[["message_id", "message", "gender","topic"]]
gender_df.hate.value_counts()
```

```python
# Downsampling

# Step 1: Count class sizes
minority_class_size = len(gender_df[gender_df.gender == True])
majority_class_size = len(gender_df[gender_df.gender == False])

# Step 2: Determine how many to drop from the majority class
downsample_size = majority_class_size - minority_class_size

# Step 3: Sample message IDs to drop
ids_to_drop = (
    gender_df[gender_df.gender == False]
    .sample(n=downsample_size, random_state=42)["message_id"]
    .to_list()
)

# Step 4: Drop those rows by message_id
gender_df = gender_df[~gender_df["message_id"].isin(ids_to_drop)]
print(len(gender_df))
gender_df.gender.value_counts()
```

```python
gender_df.to_parquet(f"{OUTPUT_PATH}/TG-weak-gender-core.parquet")
```

## HateXplain

```python
hate_df = df[["message_id", "message", "hate","topic"]]
hate_df.hate.value_counts()
```

```python
# Downsampling

# Step 1: Count class sizes
minority_class_size= len(hate_df[hate_df.hate == "hate speech"])
majority_class_size_off = len(hate_df[hate_df.hate == "offensive"])
majority_class_size_normal = len(hate_df[hate_df.hate == "normal"])

# Step 2: Determine how many to drop from the majority class
downsample_size_off = majority_class_size_off - minority_class_size
downsample_size_normal = majority_class_size_normal - minority_class_size

# Step 3: Sample message IDs to drop
ids_to_drop_off = (
    hate_df[hate_df.hate == "offensive"]
    .sample(n=downsample_size_off, random_state=42)["message_id"]
    .to_list()
)
ids_to_drop_normal = (
    hate_df[hate_df.hate == "normal"]
    .sample(n=downsample_size_normal, random_state=42)["message_id"]
    .to_list()
)
ids_to_drop = ids_to_drop_off + ids_to_drop_normal

# Step 4: Drop those rows by message_id
hate_df = hate_df[~hate_df["message_id"].isin(ids_to_drop)]
print(len(hate_df))
hate_df.hate.value_counts()
```

```python
hate_df.to_parquet(f"{OUTPUT_PATH}/TG-weak-hate-core.parquet")
```

## Sentiment

```python
senti_df = df[["message_id", "message", "senti","topic"]]
senti_df.senti.value_counts()
```

```python
# Downsampling

# Step 1: Count class sizes
minority_class_size = len(senti_df[senti_df.senti == "positive"])
majority_class_size_neg = len(senti_df[senti_df.senti == "negative"])
majority_class_size_neu = len(senti_df[senti_df.senti == "neutral"])

# Step 2: Determine how many to drop from the majority class
downsample_size_neg = majority_class_size_neg - minority_class_size
downsample_size_neu = majority_class_size_neu - minority_class_size

# Step 3: Sample message IDs to drop
ids_to_drop_neg = (
    senti_df[senti_df.senti == "negative"]
    .sample(n=downsample_size_neg, random_state=42)["message_id"]
    .to_list()
)
ids_to_drop_neu = (
    senti_df[senti_df.senti == "neutral"]
    .sample(n=downsample_size_neu, random_state=42)["message_id"]
    .to_list()
)
ids_to_drop = ids_to_drop_neg + ids_to_drop_neu

# Step 4: Drop those rows by message_id
senti_df = senti_df[~senti_df["message_id"].isin(ids_to_drop)]
print(len(senti_df))
senti_df.senti.value_counts()
```

```python
senti_df.to_parquet(f"{OUTPUT_PATH}/TG-weak-sentiment-core.parquet")
```
